## 4. Carga y revisión de los datos

```python
import pandas as pd
import os

archivo = 'libros_scraping.csv'

if os.path.exists(archivo):
    datos = pd.read_csv(archivo, encoding='utf-8-sig')
    print('Archivo cargado correctamente')
    print('Cantidad de registros:', len(datos))
    print('Columnas disponibles:', list(datos.columns))
    display(datos.head())
else:
    print('No se encontró el archivo CSV')
```
